# Fix CogVideoX import in Google Colab
This notebook repairs the orphaned/incompatible TorchAO package reported by `FqnToConfig`, then tests the CogVideoX import. It does not publish videos or change YouTube settings.

```python
import os, subprocess, sys, runpy
repo = '/content/Youtube1'
branch = 'fix/colab-torchao-cogvideox'
if not os.path.isdir(os.path.join(repo, '.git')):
    subprocess.run(['git', 'clone', '--branch', branch, 'https://github.com/farzin0369/Youtube1.git', repo], check=True)
else:
    subprocess.run(['git', '-C', repo, 'fetch', 'origin', branch], check=True)
    subprocess.run(['git', '-C', repo, 'checkout', branch], check=True)
    subprocess.run(['git', '-C', repo, 'pull', '--ff-only', 'origin', branch], check=True)
os.chdir(repo)
runpy.run_path('scripts/repair_colab_cogvideox.py', run_name='__main__')
```
