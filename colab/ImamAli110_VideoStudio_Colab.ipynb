# ImamAli110 Video Studio — Colab GPU

این نوت‌بوک یک رابط وب متن‌باز تولید ویدئو را در نشست فعلی Colab اجرا می‌کند. مبنا: [Automated Video Generator (MIT)](https://github.com/itsPremkumar/Automated-Video-Generator).

**شفافیت:** رایگان‌بودن کد به معنی GPU ابری دائمی یا تولید نامحدود ویدئوی مولد نیست. نشست Colab ممکن است قطع شود؛ این نمونه هنوز با موتور و کنترل کیفیت کانال ادغام نهایی نشده است. این نوت‌بوک چیزی را در YouTube منتشر نمی‌کند.


```python
# بررسی محیط؛ اگر GPU فعال باشد، نوع آن را نمایش می‌دهیم.
import os, subprocess, shutil
print('Python:', subprocess.check_output(['python', '--version'], text=True).strip())
print('GPU utility:', shutil.which('nvidia-smi'))
if shutil.which('nvidia-smi'):
    subprocess.run(['nvidia-smi'], check=False)
else:
    print('GPU در این نشست در دسترس نیست؛ ادامهٔ رابط وب ممکن است بدون GPU هم اجرا شود.')
print('Node:', subprocess.run(['bash','-lc','node --version'], capture_output=True, text=True).stdout.strip() or 'not installed')
```

```python
# دریافت پروژهٔ متن‌باز و نصب وابستگی‌ها در فضای موقت Colab.
import os, subprocess, pathlib
work = '/content/automated-video-generator'
if not os.path.isdir(work):
    subprocess.run(['git','clone','--depth','1','https://github.com/itsPremkumar/Automated-Video-Generator.git',work], check=True)
os.chdir(work)
if not shutil.which('node') or not shutil.which('npm'):
    subprocess.run(['bash','-lc','apt-get update -qq && apt-get install -y nodejs npm'], check=True)
print(subprocess.run(['node','--version'], capture_output=True, text=True, check=True).stdout.strip())
subprocess.run(['npm','install','--no-audit','--no-fund'], check=True)
print('نصب وابستگی‌ها تمام شد. اگر خطا دیدی، متن خطا را بررسی می‌کنیم؛ موفقیت نصب هنوز تأیید نشده تا این سلول بدون خطا تمام شود.')
```

```python
# اجرای رابط وب و ساخت پیوند دسترسی از داخل Colab.
import os, subprocess, time, urllib.request
from google.colab import output
os.chdir('/content/automated-video-generator')
env = os.environ.copy()
env['PORT'] = '3001'
server = subprocess.Popen(['npm','run','dev'], cwd=os.getcwd(), env=env, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
time.sleep(8)
if server.poll() is not None:
    print('سرور متوقف شد. خروجی:')
    print(server.stdout.read()[-12000:])
else:
    print('پیوند رابط وب (تا وقتی نشست Colab روشن باشد):')
    print(output.eval_js("google.colab.kernel.proxyPort(3001)"))
    print('اگر صفحه باز نشد، لاگ سرور را بررسی کن؛ بالا بودن فرایند به‌تنهایی سلامت رابط را ثابت نمی‌کند.')
```
